```python
import matplotlib.pyplot as plt
import numpy as np

# Data
provinces = ['Ontario', 'Quebec', 'Saskatchewan', 'Manitoba']
population = [15996989, 9030604, 1266959, 1509702]

# Convert population to numpy array FIRST
population = np.array(population)

# Base color (E33B43) in RGB normalized (0-1 range)
base_color = np.array([227/255, 59/255, 67/255])

# Normalize population values to range 0.4 to 1 for color intensity
# values_normalized = (population - min(population)) / (max(population) - min(population))
values_normalized = (population - population.min()) / (population.max() - population.min())
color_strength = 0.4 + values_normalized * 0.6  # ensures light but visible base

# Create shades
colors = [base_color * s for s in color_strength]

# Plot
plt.bar(provinces, population, color=colors)
plt.xlabel("Province")
plt.ylabel("Population")
plt.title("Canada Province Population Chart")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
provinces = ['Ontario', 'Quebec', 'Saskatchewan', 'Manitoba']
population = [15996989, 9030604, 1266959, 1509702]

# Convert population to numpy array
population = np.array(population)

# Base color (54D199) normalized
base_color = np.array([227/255, 59/255, 67/255])

# Normalize population values for color intensity
values_normalized = (population - population.min()) / (population.max() - population.min())
color_strength = 0.4 + values_normalized * 0.6

# Create shades
colors = [base_color * s for s in color_strength]

# Pie chart
plt.pie(
    population,
    labels=provinces,
    colors=colors,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Canada Province Population Distribution")
plt.axis('equal')  # ensures circle shape
plt.show()
```
