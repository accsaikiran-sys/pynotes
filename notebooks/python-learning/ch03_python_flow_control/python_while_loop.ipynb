```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

```python
number = 5

while number > 0:
    print(number)
    number -= 1
```

```python
count = 0

while True:
    if count == 3:
        break
    print(count)
    count += 1
```

```python
number = 0

while number < 5:
    number += 1
    if number == 3:
        continue
    print(number)
```

```python
i = 1

while i <= 3:
    j = 1
    while j <= 2:
        print(f"i={i}, j={j}")
        j += 1
    i += 1
```

```python
age = -1

while age < 0:
    age = int(input("Enter a positive age: "))

print("Valid age entered:", age)
```
