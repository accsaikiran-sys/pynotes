```python
temperature = 15

if temperature > 25:
    print("It's warm outside")
else:
    print("It's cold outside")
```

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 75:
    print("Grade: B")
elif score >= 60:
    print("Grade: C")
else:
    print("Grade: D")
```

```python
username = "admin"
password = "1234"

if username == "admin":
    if password == "1234":
        print("Access granted")
    else:
        print("Incorrect password")
```

```python
num = 7

result = "Even" if num % 2 == 0 else "Odd"
print(result)  # Output: Odd
```
