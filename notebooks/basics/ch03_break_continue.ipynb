```python
for number in range(1, 10):
    if number == 5:
        break
    print(number)
```

```python
for number in range(1, 6):
    if number == 3:
        continue
    print(number)
```

```python
count = 0

while True:
    if count == 3:
        break
    print(count)
    count += 1
```

```python
num = 0

while num < 5:
    num += 1
    if num == 2:
        continue
    print(num)
```

```python
for i in range(3):
    for j in range(5):
        if j == 2:
            break
        print(f"i={i}, j={j}")
```

```python
found = False

for number in range(10):
    if number == 7:
        found = True
        break

print("Found:", found)
```

```python
for n in range(5):
    if n == 6:
        break
else:
    print("Loop completed without break")
```
