```python
def greet():
    print("Hello, Python!")

greet()
```

```python
def greet_user(name):
    print(f"Hello, {name}!")

greet_user("Alice")
```

```python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)  # Output: 8
```

```python
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet()
greet("Bob")
```
