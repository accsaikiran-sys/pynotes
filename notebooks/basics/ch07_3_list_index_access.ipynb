```python
my_list = ["apple", "banana", "cherry"]

try:
    index = int(input("Enter an index to access: "))
    print(f"Item at index {index}: {my_list[index]}")

except IndexError:
    print("Index out of range!")
except ValueError:
    print("Please enter a valid integer!")
```
